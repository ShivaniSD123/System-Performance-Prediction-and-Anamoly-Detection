```python
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt

data=pd.read_csv('../system_metrics.csv')

plt.style.use('fivethirtyeight')
dates=data['timestamp']
cpu_usage=data['cpu_percent']
ram=data['ram_percent']
net_usage=data['net_bytes_per_sec']/1048576


plt.plot(dates, cpu_usage, label='CPU - Usage')
plt.plot(dates, ram, label='Ram-Gathering')
plt.plot(dates, net_usage, label='network-usage in MBs')
plt.xlabel('Date')
plt.ylabel('Actual Data')
plt.title('Data  Visualization')
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```
