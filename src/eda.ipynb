```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("../data/system_metrics.csv")

print(data.head())

print("Dataset Info:")
print(data.info())

print("Statistical Summary:")
print(data.describe())

print("Missing Values:")
print(data.isnull().sum())

print("Skewness:")
print(data.skew(numeric_only=True))

plt.figure(figsize=(10,5))
plt.plot(data['cpu_percent'])
plt.title("CPU Usage Over Time")
plt.xlabel("Time Index")
plt.ylabel("CPU %")
plt.show()

plt.figure(figsize=(10,5))
plt.plot(data['ram_percent'])
plt.title("RAM Usage Over Time")
plt.xlabel("Time Index")
plt.ylabel("RAM %")
plt.show()

data[['cpu_percent','ram_percent','net_bytes_per_sec']].hist(figsize=(10,6))
plt.suptitle("Feature Distributions")
plt.show()

plt.figure(figsize=(10,6))
sns.boxplot(data=data[['cpu_percent','ram_percent','net_bytes_per_sec']])
plt.title("Boxplot for System Metrics")
plt.show()

plt.figure(figsize=(6,4))
sns.heatmap(data.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Feature Correlation Heatmap")
plt.show()
```
