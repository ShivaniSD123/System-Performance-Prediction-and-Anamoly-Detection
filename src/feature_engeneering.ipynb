```python
import pandas as pd
#from sklearn.preprocessing import MinMaxScaler

# Load data
data = pd.read_csv('../data/system_metrics.csv')

# Rolling averages
data['cpu_roll_mean'] = data['cpu_percent'].rolling(window=10).mean()
data['ram_roll_mean'] = data['ram_percent'].rolling(window=10).mean()
data['net_roll_mean'] = data['net_bytes_per_sec'].rolling(window=10).mean()

# Rate of change
data['cpu_change'] = data['cpu_percent'].diff()
data['ram_change'] = data['ram_percent'].diff()
data['net_change'] = data['net_bytes_per_sec'].diff()

# Time features
data['hour'] = data.index.hour
data['minute'] = data.index.minute

# Remove NaN rows
data = data.dropna()

# -------- Scaling (LAST STEP) --------

#scaler = MinMaxScaler()

#scaled_columns = scaler.fit_transform(data)

#data_scaled = pd.DataFrame(
 #   scaled_columns,
  #  columns=data.columns,
   # index=data.index
#)

# Save processed dataset
data.to_csv("../data/featured_system_metrics.csv")
```
