```python
import pandas as pd
#from sklearn.preprocessing import MinMaxScaler

# Load data
data = pd.read_csv('../data/processed_system_metrics.csv')

# Convert timestamp
data['timestamp'] = pd.to_datetime(data['timestamp'])

data = data.set_index('timestamp')
# Rolling averages
data['cpu_roll_mean'] = data['cpu_percent'].rolling(window=10).mean()
data['ram_roll_mean'] = data['ram_percent'].rolling(window=10).mean()
data['net_roll_mean'] = data['net_bytes_per_sec'].rolling(window=10).mean()

# Rate of change
data['cpu_change'] = data['cpu_percent'].diff()
data['ram_change'] = data['ram_percent'].diff()
data['net_change'] = data['net_bytes_per_sec'].diff()

# Time features
data['hour'] = data.index.hour
data['minute'] = data.index.minute

#Lag Feature
lags = [1]

for lag in lags:
    
    data[f"cpu_lag{lag}"] = data["cpu_percent"].shift(lag)
    data[f"ram_lag{lag}"] = data["ram_percent"].shift(lag)
    data[f"net_lag{lag}"] = data["net_bytes_per_sec"].shift(lag)



# Remove NaN rows
data = data.dropna()
# Save processed dataset
data.to_csv("../data/featured_system_metrics.csv")
```
