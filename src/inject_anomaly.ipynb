```python
import pandas as pd

df = pd.read_csv("../data/test_data.csv")

print(df.head())
print(df.columns)

import numpy as np

def inject_anomalies(df, anomaly_ratio=0.05, random_state=42):
    np.random.seed(random_state)
    
    data = df.copy()
    n = len(data)
    
    # number of anomalies
    n_anomalies = int(n * anomaly_ratio)
    
    # random indices
    indices = np.random.choice(n, n_anomalies, replace=False)
    
    # add label column
    data['is_anomaly'] = 0
    
    # split into types
    n1 = n_anomalies // 3
    n2 = n_anomalies // 3
    n3 = n_anomalies - (n1 + n2)
    
    idx1 = indices[:n1]
    idx2 = indices[n1:n1+n2]
    idx3 = indices[n1+n2:]
    
    # ⚡ IMPORTANT: change column names if needed
    cpu_col = 'cpu_percent'
    mem_col = 'ram_percent'
    net_col = 'net_bytes_per_sec'
    
    # 1️⃣ CPU spike
    data.loc[idx1, cpu_col] = np.random.uniform(90, 100, size=len(idx1))
    
    # 2️⃣ Memory drop
    data.loc[idx2, mem_col] = np.random.uniform(5, 15, size=len(idx2))
    
    # 3️⃣ Network spike
    data.loc[idx3, net_col] += np.random.uniform(50, 100, size=len(idx3))
    
    # mark anomalies
    data.loc[indices, 'is_anomaly'] = 1
    
    print(f"Injected {n_anomalies} anomalies")
    
    return data
df_anomaly = inject_anomalies(df, anomaly_ratio=0.05)

df_anomaly.to_csv("../data/test_data_with_anomalies.csv", index=False)
```
