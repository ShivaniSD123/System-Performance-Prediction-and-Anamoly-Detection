```python
import pandas as pd

# Load your data
df = pd.read_csv("system_metrics.csv")

# Ensure timestamp is proper datetime
df['timestamp'] = pd.to_datetime(df['timestamp'])

# Sort by time
df = df.sort_values(by='timestamp')

# Remove duplicate timestamps (if any)
df = df.drop_duplicates(subset='timestamp')

# Rename columns-
df = df.rename(columns={
    'cpu_percent': 'cpu_usage_percent',
    'ram_percent': 'ram_usage_percent',
    'net_bytes_per_sec': 'network_bytes_per_sec'
})

# Optional: Round values for clean display
# df = df.round(2)

# Reset index
df = df.reset_index(drop=True)

# Save formatted dataset
df.to_csv("system_performance_dataset.csv", index=False)
```
